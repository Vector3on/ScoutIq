
# 🧠 Bloodhound VC: Weekly TFT Trainer (Colab GPU Version)
This notebook trains the Temporal Fusion Transformer (TFT) model using Colab GPU and pushes the trained model to GitHub for downstream CI/CD and inference workflows.
    

```

# 🔐 Cell 1: Load GitHub Secrets (Colab Runtime > Secrets)
from google.colab import userdata
import subprocess, os

GITHUB_TOKEN = userdata.get("GITHUB_TOKEN")
GITHUB_REPO = userdata.get("GITHUB_REPO")
GITHUB_USERNAME = userdata.get("GITHUB_USERNAME")
GITHUB_EMAIL = userdata.get("GITHUB_EMAIL")

assert all([GITHUB_TOKEN, GITHUB_REPO, GITHUB_USERNAME, GITHUB_EMAIL]), "❌ Missing one or more GitHub secrets."

# Git setup
subprocess.run(["git", "config", "--global", "user.name", GITHUB_USERNAME])
subprocess.run(["git", "config", "--global", "user.email", GITHUB_EMAIL])
    
```

```

# 📦 Cell 2: Clone GitHub Repo
repo_url = f"https://{GITHUB_TOKEN}@github.com/{GITHUB_REPO}.git"
if not os.path.exists("repo"):
    subprocess.run(["git", "clone", repo_url, "repo"])
os.chdir("repo")

# Install requirements
subprocess.run(["pip", "install", "-r", "requirements.txt"])
print("✅ Repo ready and dependencies installed.")
    
```

```

# 🚀 Cell 4: Push Trained Model to GitHub
subprocess.run(["git", "pull"])
subprocess.run(["git", "add", "artifacts/tft_model.ckpt"])
subprocess.run(["git", "commit", "-m", "🚀 update: Colab-trained TFT model pushed"])
subprocess.run(["git", "push"])
print("✅ Model pushed to GitHub.")
    
```
